removing the outliers

```python
# Remove outliers using IQR
filtered_data = df.copy()

columns_to_check = ['Open', 'High', 'Low', 'Close']

for column in columns_to_check:
    # Calculate Q1, Q3, and IQR
    Q1 = filtered_data[column].quantile(0.25)
    Q3 = filtered_data[column].quantile(0.75)
    IQR = Q3 - Q1

    # Define lower and upper bounds
    lower_bound = Q1 - 1 * IQR
    upper_bound = Q3 + 1 * IQR

    # Filter out outliers
    filtered_data = filtered_data[(filtered_data[column] >= lower_bound) & (filtered_data[column] <= upper_bound)]

print(f"Shape before removing outliers: {df.shape}")

# Display the shape of the filtered dataset
print(f"Shape after removing outliers: {filtered_data.shape}")
```

categorising

```python
# Function to calculate the number of bins using Sturges' Rule
def calculate_bins_sturges(n):
    # Sturges' formula: number of bins = log2(n) + 1
    return int(np.ceil(np.log2(n) + 1))

# Calculate the number of bins based on the dataset
n = len(df)  # Total number of records in the dataset
bins_sturges = calculate_bins_sturges(n)

# Print the recommended number of bins based on Sturges' Rule
print(f"Recommended bins (Sturges): {bins_sturges}")
```

```python
# Alternatively, calculate the IQR (Interquartile Range) for Freedman-Diaconis rule
iqr = np.percentile(df['Open'], 75) - np.percentile(df['Open'], 25)

# Calculate bin width using Freedman-Diaconis formula: bin_width = 2 * IQR / sqrt(n)
bin_width = 2 * iqr / np.sqrt(n)

# Determine the number of bins based on the Freedman-Diaconis rule
bins_fd = int((df['Open'].max() - df['Open'].min()) / bin_width)

# Print the recommended number of bins based on Freedman-Diaconis rule
print(f"Recommended bins (Freedman-Diaconis): {bins_fd}")
```

```python
# Define a function to calculate bins and labels for a given column

def create_bins_and_labels(column_name, n_bins=5):
    # Calculate the bins (ranges) based on the column's data
    min_val = df[column_name].min()
    max_val = df[column_name].max()
    bins = pd.cut(df[column_name], bins=n_bins, retbins=True)[1]
    
    # Create labels for each bin based on the ranges
    labels = [f"{round(bins[i], 2)}-{round(bins[i+1], 2)}" for i in range(len(bins)-1)]
    return bins, labels
```

```python
# Generate bins and labels for the specified columns: Open, High, Low, Close
columns_to_bin = ['Open', 'High', 'Low', 'Close']

categorised_df = df.copy()

for column in columns_to_bin:
    # Apply the create_bins_and_labels function for each column
    bins, labels = create_bins_and_labels(column)
    
    # Add a new column with the categorized data (bins) to the DataFrame
    categorised_df[f"{column}_category"] = pd.cut(df[column], bins=bins, labels=labels, include_lowest=True)

# Display the updated DataFrame
categorised_df
```
