Removing the outliers

```python
# For open close high low 

# Remove outliers using IQR
filtered_df = df.copy()

columns_to_check = ['Open', 'High', 'Low', 'Close',]

for column in columns_to_check:
    # Calculate Q1, Q3, and IQR
    Q1 = filtered_df[column].quantile(0.25)
    Q3 = filtered_df[column].quantile(0.75)
    IQR = Q3 - Q1

    # Define lower and upper bounds
    lower_bound = Q1 - 1 * IQR
    upper_bound = Q3 + 1 * IQR

    # Filter out outliers
    filtered_df = filtered_df[(filtered_df[column] >= lower_bound) & (filtered_df[column] <= upper_bound)]

print(f"Shape before removing outliers: {df.shape}")

# Display the shape of the filtered dataset
print(f"Shape after removing outliers: {filtered_df.shape}")
```

Categorizing

```python
# Function to calculate the number of bins using Sturges' Rule
def calculate_bins_sturges(n):
    # Sturges' formula: number of bins = log2(n) + 1
    return int(np.ceil(np.log2(n) + 1))

# Calculate the number of bins based on the dataset
n = len(filtered_df)  # Total number of records in the dataset
bins_sturges = calculate_bins_sturges(n)

# Print the recommended number of bins based on Sturges' Rule
print(f"Recommended bins (Sturges): {bins_sturges}")
```

```python
# Function to calculate the number of bins using Freedman-Diaconis rule
def calculate_bins_fd(column_data):
    iqr = np.percentile(column_data, 75) - np.percentile(column_data, 25)  # Calculate IQR
    bin_width = 2 * iqr / np.sqrt(len(column_data))  # Calculate bin width
    bins_fd = int((column_data.max() - column_data.min()) / bin_width) if bin_width > 0 else 1  # Avoid division by zero
    return bins_fd

# Columns to analyze
columns_to_check = ['Open', 'High', 'Low', 'Close', 'Volume']

# Calculate bins for each column
bins_fd_results = {column: calculate_bins_fd(filtered_df[column]) for column in columns_to_check}

# Print the recommended number of bins for each column
for column, bins in bins_fd_results.items():
    print(f"Recommended bins for {column} (Freedman-Diaconis): {bins}")
```

```python
def calculate_bins_scott(column_data):
    std_dev = column_data.std()  # Standard deviation
    bin_width = 3.5 * std_dev / np.sqrt(len(column_data))  # Bin width
    bins_scott = int((column_data.max() - column_data.min()) / bin_width) if bin_width > 0 else 1
    return bins_scott

# Calculate bins for each column
bins_scott_results = {column: calculate_bins_scott(filtered_df[column]) for column in columns_to_check}

# Print results
for column, bins in bins_scott_results.items():
    print(f"Recommended bins for {column} (Scott's Rule): {bins}")
```

```python
def calculate_bins_rice(n):
    return int(2 * np.cbrt(n))

# Calculate bins based on dataset size
bins_rice = calculate_bins_rice(len(filtered_df))
print(f"Recommended bins (Rice Rule): {bins_rice}")
```

Streamlines Plot

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
import warnings
import matplotlib

# Ignore deprecation warnings from Matplotlib
warnings.filterwarnings("ignore", category=matplotlib.MatplotlibDeprecationWarning)

# آماده‌سازی داده‌ها برای Streamlines Plot
x, y, z = np.meshgrid(
    np.linspace(categorized_df['Open'].min(), categorized_df['Open'].max(), 10),
    np.linspace(categorized_df['Close'].min(), categorized_df['Close'].max(), 10),
    np.linspace(categorized_df['Volume'].min(), categorized_df['Volume'].max(), 10)
)

u = np.sin(np.pi * x / x.max())
v = np.cos(np.pi * y / y.max())
w = np.sin(np.pi * z / z.max())

# ============================
# بخش 1: نمایش در پنجره جداگانه (Interactive)
# ============================

# Enable backend for opening plots in a separate window
plt.switch_backend('TkAgg')

# Create the interactive figure
fig = plt.figure(figsize=(8, 6))  # Set the initial size of the window
ax = fig.add_subplot(111, projection='3d')

# رسم جریان خطوط
ax.quiver(x, y, z, u, v, w, length=0.5, normalize=True, color='blue', alpha=0.7)

# Add title and axis labels
ax.set_title("Interactive 3D Streamlines Plot: Open vs Close vs Volume")
ax.set_xlabel("Open")
ax.set_ylabel("Close")
ax.set_zlabel("Volume")

# Display the interactive plot
plt.show()

```

```python
# ============================
# بخش 2: نمایش ثابت در سلول
# ============================

# Enable inline plotting
%matplotlib inline

# Create the static plot in the cell
plt.figure(figsize=(10, 6))
ax_static = plt.axes(projection='3d')

# رسم جریان خطوط
ax_static.quiver(x, y, z, u, v, w, length=0.5, normalize=True, color='green', alpha=0.7)

# Add title and axis labels
ax_static.set_title("Static 3D Streamlines Plot: Open vs Close vs Volume")
ax_static.set_xlabel("Open")
ax_static.set_ylabel("Close")
ax_static.set_zlabel("Volume")

# Show the static plot in the cell
plt.show()
```

Volume Rendering

```python
# Generate a 3D grid based on Open, Close, and Volume
grid_size = 30
x = np.linspace(categorized_df['Open'].min(), categorized_df['Open'].max(), grid_size)
y = np.linspace(categorized_df['Close'].min(), categorized_df['Close'].max(), grid_size)
z = np.linspace(min(categorized_df['Volume']), max(categorized_df['Volume']), grid_size)
x, y, z = np.meshgrid(x, y, z)

# Create a synthetic 3D scalar field for Volume Rendering (normalize volume values)
volume_field = np.sin((x - x.min()) * (y - y.min()) * (z - z.min()))

# Create Volume Rendering Plot
fig = go.Figure(data=go.Volume(
    x=x.flatten(),  # Flatten the grid for x-axis
    y=y.flatten(),  # Flatten the grid for y-axis
    z=z.flatten(),  # Flatten the grid for z-axis
    value=volume_field.flatten(),  # Flatten the volume data
    isomin=-1,  # Minimum iso-surface value
    isomax=1,   # Maximum iso-surface value
    opacity=0.1,  # Opacity level
    surface_count=20,  # Number of iso-surfaces
    colorscale='Viridis'  # Color scale
))

# Update layout for the plot
fig.update_layout(
    title="Volume Rendering: Open vs Close vs Volume",
    scene=dict(
        xaxis_title="Open",
        yaxis_title="Close",
        zaxis_title="Volume"
    ),
    width=900,
    height=600
)

fig.show()
```

#### PyVista

```python
import pyvista as pv
import numpy as np
import pandas as pd

# داده‌های شما
data = {
    'Open': [42.449, 42.802, 42.145, 41.574, 42.674],
    'Close': [42.423, 41.799, 41.827, 41.907, 42.406],
    'Volume': [63725, 86429, 64235, 62746, 54540],
}
categorized_df = pd.DataFrame(data)

# ایجاد یک شبکه سه‌بعدی برای Volume Rendering
grid_size = 30
x = np.linspace(categorized_df['Open'].min(), categorized_df['Open'].max(), grid_size)
y = np.linspace(categorized_df['Close'].min(), categorized_df['Close'].max(), grid_size)
z = np.linspace(min(categorized_df['Volume']), max(categorized_df['Volume']), grid_size)
x, y, z = np.meshgrid(x, y, z)

# مقادیر Volume به عنوان داده‌های حجم
volume_field = np.sin((x - x.min()) * (y - y.min()) * (z - z.min()))

# ایجاد حجم داده در PyVista
grid = pv.UniformGrid()
grid.dimensions = np.array(volume_field.shape) + 1
grid.origin = (x.min(), y.min(), z.min())
grid.spacing = (x.ptp() / grid_size, y.ptp() / grid_size, z.ptp() / grid_size)
grid.point_data["volume"] = volume_field.flatten(order="F")

# رسم نمودار Volume Rendering
plotter = pv.Plotter()
plotter.add_volume(grid, opacity="linear", cmap="viridis")
plotter.show()
```
