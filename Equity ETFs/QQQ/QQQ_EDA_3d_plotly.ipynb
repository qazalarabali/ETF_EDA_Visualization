# QQQ (Invesco QQQ Trust)

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go

%matplotlib inline
```

```python
file_path = "QQQ_categorized_data.csv"

categorized_df = pd.read_csv(file_path, delimiter=',')
```

# 3D Analysis

#### Plotly

Scatter Plot

```python
# Create a 3D scatter plot
fig = px.scatter_3d(
    categorized_df,
    x='Open',  # Set the x-axis to 'Open'
    y='Close',  # Set the y-axis to 'Close'
    z='Volume',  # Set the z-axis to 'Volume'
    color='Volume',  # Color points based on the 'Volume' column
    size='Volume',
    opacity=0.7,  # Adjust transparency of the points
    title="3D Scatter Plot: Open vs Close vs Volume"  # Set the plot title
)

# Update layout to change the size of the plot
fig.update_layout(
    width=900,  # Set the width of the plot
    height=600   # Set the height of the plot
)

# Display the plot
fig.show()
```

surface plot

```python
# Prepare the data for the surface plot
x = np.linspace(categorized_df['Open'].min(), categorized_df['Open'].max(), 30)  # Generate x-axis values
y = np.linspace(categorized_df['Close'].min(), categorized_df['Close'].max(), 30)  # Generate y-axis values
x, y = np.meshgrid(x, y)  # Create a meshgrid for x and y
z = np.sqrt(x * y)  # Calculate a derived z-value (sample function)

# Create a 3D surface plot
fig = go.Figure(data=[go.Surface(z=z, x=x, y=y, colorscale='Viridis')])  # Set the color scale to 'Viridis'
fig.update_layout(
    title="3D Surface Plot: Open vs Close vs Derived Z",  # Set the plot title
    scene=dict(
        xaxis_title='Open',  # Label for the x-axis
        yaxis_title='Close',  # Label for the y-axis
        zaxis_title='Derived Z'  # Label for the z-axis
    )
)

# Update layout to change the size of the plot
fig.update_layout(
    width=900,  # Set the width of the plot
    height=600   # Set the height of the plot
)

# Display the plot
fig.show()
```

3D contour plot

```python
# Create a 3D contour plot
fig = go.Figure(data=[go.Contour(
    z=np.sqrt(x * y),  # Calculate a derived z-value (sample function)
    x=np.linspace(categorized_df['Open'].min(), categorized_df['Open'].max(), 30),  # x-axis values
    y=np.linspace(categorized_df['Close'].min(), categorized_df['Close'].max(), 30),  # y-axis values
    colorscale='Viridis',  # Set the color scale to 'Viridis'
    contours=dict(start=0, end=100, size=5)  # Define the contour levels
)])

fig.update_layout(
    title="3D Contour Plot: Open vs Close vs Derived Z",  # Set the plot title
    scene=dict(
        xaxis_title='Open',  # Label for the x-axis
        yaxis_title='Close',  # Label for the y-axis
        zaxis_title='Derived Z'  # Label for the z-axis
    )
)

# Update layout to change the size of the plot
fig.update_layout(
    width=900,  # Set the width of the plot
    height=600   # Set the height of the plot
)

# Display the plot
fig.show()
```

streamtube plot

```python
# Prepare data for the streamtube plot
x, y, z = np.meshgrid(
    np.linspace(categorized_df['Open'].min(), categorized_df['Open'].max(), 10),  # x-axis values
    np.linspace(categorized_df['Close'].min(), categorized_df['Close'].max(), 10),  # y-axis values
    np.linspace(categorized_df['Volume'].min(), categorized_df['Volume'].max(), 10)  # z-axis values
)

u = np.sin(np.pi * x / x.max())  # x-direction velocity
v = np.cos(np.pi * y / y.max())  # y-direction velocity
w = np.sin(np.pi * z / z.max())  # z-direction velocity

# Create a streamtube plot
fig = go.Figure(data=go.Streamtube(
    x=x.flatten(),
    y=y.flatten(),
    z=z.flatten(),
    u=u.flatten(),
    v=v.flatten(),
    w=w.flatten(),
    colorscale='Viridis',  # Set the color scale
    showscale=True  # Display the color scale
))

fig.update_layout(
    title="3D Streamtube Plot: Open vs Close vs Volume",  # Set the plot title
    scene=dict(
        xaxis_title='Open',  # Label for the x-axis
        yaxis_title='Close',  # Label for the y-axis
        zaxis_title='Volume'  # Label for the z-axis
    )
)

# Update layout to change the size of the plot
fig.update_layout(
    width=900,  # Set the width of the plot
    height=600   # Set the height of the plot
)

# Display the plot
fig.show()
```

3D heatmap

```python
# Prepare data for the heatmap
x = np.linspace(categorized_df['Open'].min(), categorized_df['Open'].max(), 30)  # x-axis values
y = np.linspace(categorized_df['Close'].min(), categorized_df['Close'].max(), 30)  # y-axis values
x, y = np.meshgrid(x, y)  # Create a meshgrid for x and y
z = np.sqrt(x * y)  # Calculate derived z-values

# Create a heatmap
fig = go.Figure(data=go.Heatmap(
    z=z,  # Set z-values
    x=x[0],  # Set x-axis values
    y=y[:, 0],  # Set y-axis values
    colorscale='Jet'  # Set the color scale
))

# Viridis, Inferno, Jet
# colorscale: Viridis, Cividis, Plasma, Inferno, Magma, Jet, Rainbow, Blues, Greens

fig.update_layout(
    title="3D Heatmap: Open vs Close",  # Set the plot title
    xaxis_title="Open",  # Label for the x-axis
    yaxis_title="Close"  # Label for the y-axis
)

# Update layout to change the size of the plot
fig.update_layout(
    width=900,  # Set the width of the plot
    height=600   # Set the height of the plot
)

# Display the plot
fig.show()
```

3D Surface bar plot

```python
# Reduce the number of points for grid creation
sample_size = 100  # Number of points to sample
sampled_df = categorized_df.sample(n=sample_size, random_state=42)  # Random sampling

# Create a grid for Open and Close
x = np.linspace(sampled_df['Open'].min(), sampled_df['Open'].max(), 30)
y = np.linspace(sampled_df['Close'].min(), sampled_df['Close'].max(), 30)
x, y = np.meshgrid(x, y)

# Map Volume to the z matrix
z = np.histogram2d(
    sampled_df['Open'], 
    sampled_df['Close'], 
    bins=[30, 30], 
    weights=sampled_df['Volume']
)[0]

# Replace NaN values with zeros
z = np.nan_to_num(z)

# Create a 3D Surface Plot
fig = go.Figure(data=[go.Surface(z=z, x=x, y=y, colorscale='Viridis')])

fig.update_layout(
    title="3D Surface Plot: Open vs Close vs Volume (Sampling)",
    scene=dict(
        xaxis_title="Open",
        yaxis_title="Close",
        zaxis_title="Volume"
    ),
    width=900,
    height=600
)

fig.show()
```

3D Surface Plot

```python
categorized_df['Open_bin'] = pd.cut(categorized_df['Open'], bins=30, labels=False)
categorized_df['Close_bin'] = pd.cut(categorized_df['Close'], bins=30, labels=False)

grouped = categorized_df.groupby(['Open_bin', 'Close_bin'])['Volume'].mean().unstack(fill_value=0)

x = np.linspace(categorized_df['Open'].min(), categorized_df['Open'].max(), 30)
y = np.linspace(categorized_df['Close'].min(), categorized_df['Close'].max(), 30)
x, y = np.meshgrid(x, y)
z = grouped.values

fig = go.Figure(data=[go.Surface(z=z, x=x, y=y, colorscale='Viridis')])
fig.update_layout(
    title="3D Surface Plot (Grouped Averages)",
    scene=dict(
        xaxis_title="Open",
        yaxis_title="Close",
        zaxis_title="Volume"
    ),
    width=900,
    height=600
)

fig.show()
```

Volume Rendering

```python
# Generate a 3D grid based on Open, Close, and Volume
grid_size = 30
x = np.linspace(categorized_df['Open'].min(), categorized_df['Open'].max(), grid_size)
y = np.linspace(categorized_df['Close'].min(), categorized_df['Close'].max(), grid_size)
z = np.linspace(min(categorized_df['Volume']), max(categorized_df['Volume']), grid_size)
x, y, z = np.meshgrid(x, y, z)

# Create a synthetic 3D scalar field for Volume Rendering (normalize volume values)
volume_field = np.sin((x - x.min()) * (y - y.min()) * (z - z.min()))

# Create Volume Rendering Plot
fig = go.Figure(data=go.Volume(
    x=x.flatten(),  # Flatten the grid for x-axis
    y=y.flatten(),  # Flatten the grid for y-axis
    z=z.flatten(),  # Flatten the grid for z-axis
    value=volume_field.flatten(),  # Flatten the volume data
    isomin=-1,  # Minimum iso-surface value
    isomax=1,   # Maximum iso-surface value
    opacity=0.1,  # Opacity level
    surface_count=20,  # Number of iso-surfaces
    colorscale='Viridis'  # Color scale
))

# Update layout for the plot
fig.update_layout(
    title="Volume Rendering: Open vs Close vs Volume",
    scene=dict(
        xaxis_title="Open",
        yaxis_title="Close",
        zaxis_title="Volume"
    ),
    width=900,
    height=600
)

fig.show()
```
